## Project goal
This MVP creates a functional baseline Tic-Tac-Toe game. Two simulated players choose random valid moves. The program can play one visible round, identify a winner or draw, record the number of moves, and simulate 100 rounds to generate data.

## 1. Import the tools used in the MVP

```python
import random
from collections import Counter
import matplotlib.pyplot as plt

# A fixed seed makes the demonstration reproducible.
random.seed(42)
```

## 2. Game functions

```python
WINNING_LINES = [
    (0, 1, 2), (3, 4, 5), (6, 7, 8),  # rows
    (0, 3, 6), (1, 4, 7), (2, 5, 8),  # columns
    (0, 4, 8), (2, 4, 6)              # diagonals
]

def display_board(board):
    """Display the current 3-by-3 board."""
    for row in range(0, 9, 3):
        print(f' {board[row]} | {board[row + 1]} | {board[row + 2]} ')
        if row < 6:
            print('---+---+---')

def check_winner(board):
    """Return X or O after a win, Draw if full, or None otherwise."""
    for a, b, c in WINNING_LINES:
        if board[a] != ' ' and board[a] == board[b] == board[c]:
            return board[a]
    if ' ' not in board:
        return 'Draw'
    return None

def play_game(show_game=False):
    """Simulate one game using random valid moves."""
    board = [' '] * 9
    player = 'X'
    moves = 0

    while True:
        available_positions = [i for i, value in enumerate(board) if value == ' ']
        position = random.choice(available_positions)
        board[position] = player
        moves += 1

        if show_game:
            print(f'Move {moves}: Player {player} selects position {position + 1}')
            display_board(board)
            print()

        result = check_winner(board)
        if result is not None:
            if show_game:
                print(f'Final result: {result}')
            return result, moves

        player = 'O' if player == 'X' else 'X'
```

## 3. Demonstrate one round of gameplay
The following cell displays every move in one complete game. Board positions are selected randomly from the valid unoccupied spaces.

```python
one_game_result, one_game_moves = play_game(show_game=True)
```

## 4. Simulate 100 rounds
This simulation records each result and the number of moves required to finish each game.

```python
results = []
move_counts = []

for game_number in range(100):
    result, moves = play_game()
    results.append(result)
    move_counts.append(moves)

result_counts = Counter(results)
print('Results after 100 games:')
print(f"X wins: {result_counts['X']}")
print(f"O wins: {result_counts['O']}")
print(f"Draws: {result_counts['Draw']}")
print(f"Average moves per game: {sum(move_counts) / len(move_counts):.2f}")
```

## 5. Visualize the results

```python
labels = ['X Wins', 'O Wins', 'Draws']
values = [result_counts['X'], result_counts['O'], result_counts['Draw']]
colors = ['#4C78A8', '#F58518', '#9E9E9E']

plt.figure(figsize=(8, 5))
bars = plt.bar(labels, values, color=colors)
plt.title('Results of 100 Simulated Tic-Tac-Toe Games')
plt.xlabel('Game Result')
plt.ylabel('Number of Games')
plt.ylim(0, max(values) + 10)
for bar, value in zip(bars, values):
    plt.text(bar.get_x() + bar.get_width()/2, value + 1, str(value),
             ha='center', fontweight='bold')
plt.show()
```
